## Plain-language result
See section 7 of the methods doc. In short: Crescent City's landed value during the closure was in line with its counterfactual (SC +3%, DiD +10%), with no sign of a shift into the South-OR (Bandon–Brookings) unit; but year-to-year noise is large (placebo gaps −84% to +161%), so the permutation tests cannot statistically rule out even a large loss, and Crescent City's shortfall against Humboldt Bay alone was already present the season before the closure.
